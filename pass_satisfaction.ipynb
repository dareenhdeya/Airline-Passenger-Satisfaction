## Data Overview

this dataset is about airline passenger satisfaction, the target variable is satisfaction which has two classes:
- Satisfied
- Neutral or Dissatisfied

The dataset contains passenger information, flight information, service ratings, delay information, and the target variable


### explaining columns
- Passenger Information

    - **Gender:** passenger's gender (Male or Female)
    - **Customer Type:** indicates whether the passenger is a loyal customer or a disloyal customer
    - **Age:** passenger's age
    - **Type of Travel:** purpose of travel (Business or Personal)
    - **Class:** travel class (Business, Eco, Eco Plus)
    - **Flight Distance:** total flight distance in kilometers

- Service Ratings

        most of these features are customer ratings ranging from 1 to 5 where higher values indicate better satisfaction and zero means N/A

    - **Inflight wifi service:** rating of the onboard wifi quality
    - **Departure/Arrival time convenient:** rating of how suitable the flight's scheduled departure and arrival times were
    - **Ease of Online booking:** rating of how easy it was to book the flight online
    - **Gate location:** rating of the gate location and how easy it was to reach
    - **Food and drink:** rating of food and drinks
    - **Online boarding:** rating of the online check-in and boarding process
    - **Seat comfort:** rating of seat comfort
    - **Inflight entertainment:** rating of entertainment provided during the flight
    - **On-board service:** rating of the cabin crew's overall service
    - **Leg room service:** rating of legroom space
    - **Baggage handling:** rating of how well the airline handled passenger luggage
    - **Check-in service:** rating of the airport check-in experience
    - **Inflight service:** rating of the service provided during the flight
    - **Cleanliness:** rating of aircraft cleanliness

- Delay Information

    - **Departure Delay in Minutes:** number of minutes the flight was delayed before takeoff
    - **Arrival Delay in Minutes:** number of minutes the flight was delayed when it arrived

- target variable
    - **satisfaction** :  indicates whether the passenger was satisfied or neutral/dissatisfied with the overall flight experience

## Checking Outliers

```python
cols_to_check=["Flight Distance", "Departure Delay in Minutes", "Arrival Delay in Minutes"]
for col in cols_to_check:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1-1.5*IQR
    upper_bound  = Q3+1.5*IQR

    outliers = (df[col]>upper_bound) | (df[col]<lower_bound)
    print(f"{col}: number of outliers={outliers.sum()}, lb={lower_bound} ub={upper_bound} ")
```

## Handling Outliers

```python
cols_to_transform = ["Flight Distance", "Departure Delay in Minutes", "Arrival Delay in Minutes"]

#train data
for col in cols_to_transform:
    df[f"{col}_log"] = np.log1p(df[col])

#test data
for col in cols_to_transform:
    df_test[f"{col}_log"] = np.log1p(df_test[col])
```

## feature engineering

```python
def categorize_delay(min):
    if min == 0:
        return "On Time"
    elif min <= 30:
        return "Minor Delay"
    elif min <= 90:
        return "Moderate Delay"
    else:
        return "Major Delay"

# only needed for eda -> apply only on train data
df["Arrival Delay category"] = df["Arrival Delay in Minutes"].apply(categorize_delay)
df["Departure Delay category"] = df["Departure Delay in Minutes"].apply(categorize_delay)
```

```python
def age_group(age):
    if age <= 18:
        return "Young"
    elif age <= 25:
        return "Young Adult"
    elif age <= 50:
        return "Adult"
    else:
        return "Senior"
    
# only needed for eda -> apply only on train data
df["age group"] = df["Age"].apply(age_group)
```

## Preprocessing

### define features and target variable

```python
X_train = df.drop(["satisfaction", "age group", "Arrival Delay category", "Departure Delay category", "Flight Distance", "Departure Delay in Minutes", "Arrival Delay in Minutes"] ,axis=1)
y_train = df["satisfaction"]

X_test = df_test.drop(["satisfaction", "Flight Distance", "Departure Delay in Minutes", "Arrival Delay in Minutes"] ,axis=1)
y_test = df_test["satisfaction"]
```

### data encoding

```python
features = ["Type of Travel", "Customer Type", "Gender"]

for f in features:
    le = LabelEncoder()                          
    X_train[f] = le.fit_transform(X_train[f])   
    X_test[f] = le.transform(X_test[f]) 
```

```python
X_train = pd.get_dummies(X_train, columns=["Class"], dtype=int)

X_test = pd.get_dummies(X_test, columns=["Class"], dtype=int)
```

```python
le_target = LabelEncoder()

y_train = le_target.fit_transform(y_train)

y_test = le_target.transform(y_test)
```

### data scaling

```python
scaler = StandardScaler()

num_cols = ['Age', "Flight Distance_log", "Departure Delay in Minutes_log", "Arrival Delay in Minutes_log"]
X_train[num_cols] = scaler.fit_transform(X_train[num_cols])

X_test[num_cols] = scaler.transform(X_test[num_cols])
```

## ML Models

### logistic regression

```python
lr = LogisticRegression()
lr.fit(X_train, y_train)
```

```python
y_pred_lr = lr.predict(X_test)
```

```python
print(accuracy_score(y_test, y_pred_lr))
print(classification_report(y_test, y_pred_lr))
print(confusion_matrix(y_test, y_pred_lr))
```

### KNN

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
y_pred_knn = knn.predict(X_test)
```

```python
print(accuracy_score(y_test, y_pred_knn))
print(classification_report(y_test, y_pred_knn))
print(confusion_matrix(y_test, y_pred_knn))
```

### SVC

```python
svc = SVC()
svc.fit(X_train, y_train)
```

```python
y_pred_svc = svc.predict(X_test)
```

```python
print(accuracy_score(y_test, y_pred_svc))
print(classification_report(y_test, y_pred_svc))
print(confusion_matrix(y_test, y_pred_svc))
```
